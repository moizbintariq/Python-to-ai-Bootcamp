# Day 2 - NumPy Basics
NumPy works with **arrays**. An array is like a list, but we can do maths on all the values at once.

```python
import numpy as np
```

## Creating an array

```python
marks = np.array([12, 5, 8, 20, 3, 15])
print(marks)
print(type(marks))
```

## sum, mean, max, min

```python
print("Total   :", np.sum(marks))
print("Average :", np.mean(marks))
print("Highest :", np.max(marks))
print("Lowest  :", np.min(marks))
print("Count   :", len(marks))
```

## Maths on the whole array
With a normal list `*2` repeats the list, but with a NumPy array it multiplies every value.

```python
normal_list = [1, 2, 3]
print(normal_list * 2)

arr = np.array([1, 2, 3])
print(arr * 2)
print(arr + 10)
```

## Conditions on arrays

```python
print(marks > 10)
print(marks[marks > 10])
```

## 2D array (rows and columns)

```python
table = np.array([[1, 2, 3],
                  [4, 5, 6]])
print(table)
print("Shape :", table.shape)
print("Total :", table.sum())
print("Column totals:", table.sum(axis=0))
print("Row totals   :", table.sum(axis=1))
```

## Useful helpers

```python
print(np.arange(0, 10, 2))
print(np.zeros(3))
print(np.ones((2, 2)))
```
