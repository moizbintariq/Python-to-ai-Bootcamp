## 6. Filtering rows with conditions
Use `&` for AND and `|` for OR, and put every condition inside its own brackets.

```python
females = df.loc[df["Sex"] == "female"]
print("Female passengers:", len(females))
```

```python
older = df.loc[df["Age"] > 60]
print("Passengers older than 60:", len(older))
```

```python
female_survivors = df.loc[(df["Sex"] == "female") & (df["Survived"] == 1)]
print("Females who survived:", len(female_survivors))
```

# Custom Questions
Each question is solved in small steps: filter the group, count, then calculate the percentage.

## Question 1
**How many girls (female, under 18 years old) travelling in 3rd class survived?**

```python
girls = df.loc[(df["Sex"] == "female") & (df["Age"] < 18) & (df["Pclass"] == 3)]
print("Girls under 18 in 3rd class:", len(girls))

girls_survived = girls.loc[girls["Survived"] == 1]
print("Survived:", len(girls_survived))

percent = len(girls_survived) / len(girls) * 100
print("Survival rate:", round(percent, 1), "%")
```

**Finding:** 35 girls under 18 travelled in 3rd class and 19 of them survived (54.3%).

## Question 2
**How many passengers travelling alone (no family on board) in 2nd class survived?**

Alone means `SibSp == 0` and `Parch == 0`.

```python
alone = df.loc[(df["SibSp"] == 0) & (df["Parch"] == 0) & (df["Pclass"] == 2)]
print("Alone in 2nd class:", len(alone))

alone_survived = alone.loc[alone["Survived"] == 1]
print("Survived:", len(alone_survived))

print("Survival rate:", round(len(alone_survived) / len(alone) * 100, 1), "%")
```

```python
males = alone.loc[alone["Sex"] == "male"]
females = alone.loc[alone["Sex"] == "female"]

print("Males  :", len(males), "total,", len(males.loc[males["Survived"] == 1]), "survived")
print("Females:", len(females), "total,", len(females.loc[females["Survived"] == 1]), "survived")
```

**Finding:** 104 passengers travelled alone in 2nd class and 36 survived (34.6%). Among them 29 of 32 women survived but only 7 of 72 men.

## Question 3
**Did children (under 10) survive more than adults (18 and older)?**

```python
children = df.loc[df["Age"] < 10]
adults = df.loc[df["Age"] >= 18]

children_survived = children.loc[children["Survived"] == 1]
adults_survived = adults.loc[adults["Survived"] == 1]

print("Children:", len(children), "| survived:", len(children_survived))
print("Adults  :", len(adults), "| survived:", len(adults_survived))

print("Children survival rate:", round(len(children_survived) / len(children) * 100, 1), "%")
print("Adults survival rate  :", round(len(adults_survived) / len(adults) * 100, 1), "%")
```

**Finding:** 38 of 62 children survived (61.3%) compared with 229 of 601 adults (38.1%). Children had a clearly better chance, which matches the "women and children first" rule.

## Question 4
**How many passengers paid a fare above 100, how many of them survived, and who paid the highest fare?**

```python
rich = df.loc[df["Fare"] > 100]
print("Fare above 100:", len(rich))

rich_survived = rich.loc[rich["Survived"] == 1]
print("Survived:", len(rich_survived))
print("Survival rate:", round(len(rich_survived) / len(rich) * 100, 1), "%")

print()
print(rich["Pclass"].value_counts())
```

```python
df.sort_values("Fare", ascending=False)[["Name", "Pclass", "Fare", "Survived"]].head(3)
```

**Finding:** 53 passengers paid more than 100 and 39 of them survived (73.6%). 53 of the 53 were in 1st class. The highest fare was about 512.33.

## Bonus Question
**What was the average fare paid at each port?**

```python
avg_fare = df.groupby("Embarked")["Fare"].mean()
print(avg_fare.round(2))
```

**Finding:** Cherbourg (C) had the highest average fare (about 59.95), followed by Southampton (27.08) and Queenstown (13.28). Cherbourg had the biggest share of 1st-class passengers.

## Summary
- Gender, age and ticket class strongly affected survival.
- Use `loc` with conditions to build a group, `len()` to count it, and divide to get a percentage.
- Always check for missing values before filtering, because missing `Age` values are left out of age conditions.
